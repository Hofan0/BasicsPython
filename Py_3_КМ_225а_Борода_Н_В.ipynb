```python
import re

def format_initials(full_name: str) -> str:
    parts = full_name.strip().split()
    if len(parts) != 3:
        raise ValueError("Потрібно ввести три слова")
    first_name, patronymic, last_name = parts
    return f"{first_name[0].upper()}. {patronymic[0].upper()}. {last_name}"

assert format_initials("Володимир Олександрович Зеленський") == "В. О. Зеленський"
assert format_initials("Тарас Григорович Шевченко") == "Т. Г. Шевченко"
assert format_initials("Іван Якович Франко") == "І. Я. Франко"
assert format_initials("Леся Петрівна Українка") == "Л. П. Українка"
assert format_initials("Олександр Сергійович Пушкін") == "О. С. Пушкін"

def change_key(d: dict, old_key, new_key) -> dict:
    new_dict = dict(d)
    if old_key in new_dict:
        new_dict[new_key] = new_dict.pop(old_key)
    return new_dict

sample_dict = {'1': 1, '2': 2, '3': 3, 'total': 55}

assert change_key(sample_dict, 'total', 'sum') == {'1': 1, '2': 2, '3': 3, 'sum': 55}
assert change_key({'a': 100, 'b': 200}, 'a', 'alpha') == {'alpha': 100, 'b': 200}
assert change_key({'x': 1}, 'y', 'z') == {'x': 1}
assert change_key({}, 'old', 'new') == {}
assert change_key({1: 'one', 2: 'two'}, 1, 100) == {100: 'one', 2: 'two'}

with open("text.txt", "r", encoding="utf-8") as file:
    text = file.read()

cleaned_text = re.sub(r"'s\b", "", text)
cleaned_text = re.sub(r"[^\w\s]", "", cleaned_text)
words = cleaned_text.lower().split()

word_counts = {}
for word in words:
    word_counts[word] = word_counts.get(word, 0) + 1

most_frequent_word = max(word_counts, key=word_counts.get)

sorted_by_keys = dict(sorted(word_counts.items()))

sorted_by_values = dict(sorted(word_counts.items(), key=lambda item: item[1]))

print("Частота слів (перші 10):", list(word_counts.items())[:10])
print("Найуживаніше слово:", most_frequent_word, ":", word_counts[most_frequent_word])
print("Сортування за ключами (перші 5):", list(sorted_by_keys.items())[:5])
print("Сортування за значеннями (перші 5):", list(sorted_by_values.items())[:5])
```
