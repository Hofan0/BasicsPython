```python
def is_prime(n: int) -> bool:
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def task_9(input_str: str = None):
    if input_str is None:
        input_str = input()

    tokens = input_str.strip().split()
    numbers = []

    for token in tokens:
        try:
            numbers.append(int(token))
        except ValueError:
            return "Помилка. У списку присутнє нечислове значення"

    non_primes = [num for num in numbers if not is_prime(num)]

    non_primes.sort(reverse=True)

    return non_primes

print("Тест 1:")
print("Ввод: 3 12 15 7 11 35 18")
print("Out:", task_9("3 12 15 7 11 35 18"))

print("\nТест 2:")
print("Ввод: 0 34 2 1 15 22 4 23 6")
print("Out:", task_9("0 34 2 1 15 22 4 23 6"))

print("\nТест 3 (Перевірка помилки некоректного введення):")
print("Ввод: 3 4 23 5 23 str 3 4")
print("Out:", task_9("3 4 23 5 23 str 3 4"))
```
