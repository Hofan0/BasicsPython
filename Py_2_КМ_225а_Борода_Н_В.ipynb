```python
def task_1_var4(n: int, lst: list):
    if len(lst) < 3 or n <= 2:
        return

    new_lst = list(lst)
    min_val = min(new_lst)
    max_val = max(new_lst)

    min_idx = new_lst.index(min_val)
    max_idx = new_lst.index(max_val)

    new_lst[min_idx], new_lst[max_idx] = new_lst[max_idx], new_lst[min_idx]

    return new_lst

def task_2_var4(lst: list):
    if not lst:
        return []

    new_lst = list(lst)
    min_val = min(new_lst)
    min_idx = new_lst.index(min_val)

    new_lst.pop(min_idx)
    return new_lst

def task_3_var4(lst: list):
    if not lst or len(lst) == 1:
        return list(lst)

    new_lst = list(lst)
    first_element = new_lst.pop(0)
    new_lst.append(first_element)

    return new_lst

print("--- ТЕСТИ ЗАВДАННЯ 1 ---")
print(task_1_var4(4, [1, 4, 2, 5]))
print(task_1_var4(3, [2, 6, 8]))
print(task_1_var4(5, [5, 25, 30, 20, 15]))

print("\n--- ТЕСТИ ЗАВДАННЯ 2 ---")
original_1 = [1, 2, 3, 4, 5]
print("Результат:", task_2_var4(original_1))
print("Початковий список не змінився:", original_1)
print(task_2_var4([5, 3, 2, 1, 4]))
print(task_2_var4([2, 2, 1, 2, 1]))

print("\n--- ТЕСТИ ЗАВДАННЯ 3 ---")
print(task_3_var4([1, 2, 3, 4]))
print(task_3_var4([1]))
print(task_3_var4([]))
```
